```python
def retrieve_evidence(query, documents, top_k=3):
    query_words = set(query.lower().split())

    scored_documents = []

    for doc in documents:
        text = doc["text"].lower()

        score = sum(
            1 for word in query_words
            if word in text
        )

        scored_documents.append(
            (score, doc)
        )

    scored_documents.sort(
        key=lambda x: x[0],
        reverse=True
    )

    return [
        doc for score, doc in scored_documents[:top_k]
        if score > 0
    ]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def semantic_retrieve(query, documents, embeddings, model, top_k=3):

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    similarities = cosine_similarity(
        query_embedding,
        embeddings
    )[0]

    ranked_indices = similarities.argsort()[::-1]

    results = []

    for index in ranked_indices[:top_k]:
        results.append({
            "attack": documents[index]["attack"],
            "severity": documents[index]["severity"],
            "score": float(similarities[index]),
            "text": documents[index]["text"]
        })

    return results
```

```python
import json
import os

# Load authoritative sources
knowledge_path = "/content/drive/MyDrive/Cybersecurity_Project/evidence/knowledge/authoritative_sources.json"

with open(knowledge_path, "r", encoding="utf-8") as f:
    authoritative_data = json.load(f)

sources = authoritative_data["sources"]

# Create structured evidence documents
authoritative_documents = []

for source in sources:

    attack_family = source["attack_family"]
    source_name = source["source_name"]
    source_type = source["source_type"]

    technique_id = source.get("technique_id", "")
    technique_name = source.get("technique_name", "")

    evidence = source.get("evidence", {})

    text_parts = []

    # Basic identity
    text_parts.append(f"Attack Family: {attack_family}")
    text_parts.append(f"Source: {source_name}")
    text_parts.append(f"Source Type: {source_type}")

    if technique_id:
        text_parts.append(f"ATT&CK ID: {technique_id}")

    if technique_name:
        text_parts.append(f"Technique: {technique_name}")

    # Description
    if "description" in evidence:
        text_parts.append(
            f"Description: {evidence['description']}"
        )

    # Attack categories
    if "attack_categories" in evidence:
        text_parts.append("Attack Categories:")

        for category in evidence["attack_categories"]:
            text_parts.append(
                f"- {category['name']}: {category['description']}"
            )

    # Indicators
    if "indicators" in evidence:
        text_parts.append("Indicators:")

        for indicator in evidence["indicators"]:
            text_parts.append(f"- {indicator}")

    # Detection guidance
    detection = evidence.get(
        "detection_guidance",
        evidence.get("investigation_guidance", [])
    )

    if detection:
        text_parts.append("Detection / Investigation Guidance:")

        for item in detection:
            text_parts.append(f"- {item}")

    # Mitigation
    mitigation = evidence.get(
        "mitigation_guidance",
        evidence.get("mitigation", [])
    )

    if mitigation:
        text_parts.append("Mitigation Guidance:")

        for item in mitigation:
            text_parts.append(f"- {item}")

    # Source URL
    if source.get("source_url"):
        text_parts.append(
            f"Source URL: {source['source_url']}"
        )

    document_text = "\n".join(text_parts)

    authoritative_documents.append({
        "document_id": len(authoritative_documents),
        "attack_family": attack_family,
        "source_name": source_name,
        "source_url": source.get("source_url", ""),
        "technique_id": technique_id,
        "technique_name": technique_name,
        "text": document_text
    })

print("Created documents:", len(authoritative_documents))

for doc in authoritative_documents:
    print(
        f"{doc['document_id']} → "
        f"{doc['attack_family']} → "
        f"{doc['source_name']}"
    )
```

```python
import numpy as np

def predict_attack_family(X, model_type="xgboost"):

    if model_type.lower() == "random_forest":

        prediction = rf_model.predict(X)

        # RF already returns string labels
        attack_family = prediction[0]

        probabilities = rf_model.predict_proba(X)[0]
        confidence = float(np.max(probabilities))

    elif model_type.lower() == "xgboost":

        prediction = xgb_model.predict(X)

        # Convert numpy output to integer class
        class_id = int(np.asarray(prediction).ravel()[0])

        attack_family = label_encoder.inverse_transform(
            [class_id]
        )[0]

        probabilities = xgb_model.predict_proba(X)[0]
        confidence = float(np.max(probabilities))

    else:
        raise ValueError(
            "model_type must be 'random_forest' or 'xgboost'"
        )

    return {
        "attack_family": attack_family,
        "confidence": confidence
    }
```

```python
def get_evidence_for_attack(attack_family):

    matches = [
        doc
        for doc in authoritative_documents
        if doc["attack_family"].lower() == attack_family.lower()
    ]

    if not matches:
        return None

    return matches[0]
```

```python
predicted_attack = rf_result["attack_family"]

evidence = get_evidence_for_attack(
    predicted_attack
)

print("Predicted Attack:", predicted_attack)

if evidence:
    print("\nEvidence found!")
    print("Attack Family:", evidence["attack_family"])
    print("Source:", evidence["source_name"])
    print("Technique:", evidence["technique_name"])
else:
    print("\nNo evidence found.")
```

```python
def generate_incident_analysis(
    prediction_result,
    model_name="Random Forest"
):

    attack_family = prediction_result["attack_family"]
    confidence = prediction_result["confidence"]

    evidence = get_evidence_for_attack(
        attack_family
    )

    if evidence is None:
        return {
            "status": "No evidence found",
            "attack_family": attack_family,
            "confidence": confidence
        }

    # Find the original source entry
    source_entry = next(
        (
            source
            for source in authoritative_data["sources"]
            if source["attack_family"].lower()
            == attack_family.lower()
        ),
        None
    )

    result = {
        "model": model_name,
        "attack_family": attack_family,
        "confidence": confidence,
        "source": evidence["source_name"],
        "source_url": evidence["source_url"],
        "technique_id": evidence.get("technique_id", ""),
        "technique_name": evidence.get("technique_name", ""),
        "description": "",
        "indicators": [],
        "investigation": [],
        "mitigation": []
    }

    if source_entry:

        source_evidence = source_entry.get(
            "evidence", {}
        )

        result["description"] = source_evidence.get(
            "description",
            ""
        )

        result["indicators"] = source_evidence.get(
            "indicators",
            []
        )

        result["investigation"] = source_evidence.get(
            "detection_guidance",
            []
        )

        result["mitigation"] = source_evidence.get(
            "mitigation_guidance",
            []
        )

    return result
```

```python
def analyze_network_flow(X, model_type="random_forest"):

    # -----------------------------------------
    # 1. ML prediction
    # -----------------------------------------
    prediction = predict_attack_family(
        X,
        model_type=model_type
    )

    attack_family = prediction["attack_family"]
    confidence = prediction["confidence"]

    # -----------------------------------------
    # 2. Retrieve authoritative evidence
    # -----------------------------------------
    evidence = get_evidence_for_attack(
        attack_family
    )

    # -----------------------------------------
    # 3. Build result
    # -----------------------------------------
    if evidence is None:

        return {
            "model": model_type,
            "attack_family": attack_family,
            "confidence": confidence,
            "evidence_found": False
        }

    # Find original authoritative source
    source_entry = next(
        (
            source
            for source in authoritative_data["sources"]
            if source["attack_family"].lower()
            == attack_family.lower()
        ),
        None
    )

    source_evidence = {}

    if source_entry:
        source_evidence = source_entry.get(
            "evidence",
            {}
        )

    result = {
        "model": model_type,
        "attack_family": attack_family,
        "confidence": confidence,
        "evidence_found": True,
        "source": evidence["source_name"],
        "source_url": evidence["source_url"],
        "technique_id": evidence.get(
            "technique_id",
            ""
        ),
        "technique_name": evidence.get(
            "technique_name",
            ""
        ),
        "description": source_evidence.get(
            "description",
            ""
        ),
        "indicators": source_evidence.get(
            "indicators",
            []
        ),
        "investigation": source_evidence.get(
            "detection_guidance",
            []
        ),
        "mitigation": source_evidence.get(
            "mitigation_guidance",
            []
        )
    }

    return result
```

```python
def compare_models(X):

    rf_result = predict_attack_family(
        X,
        model_type="random_forest"
    )

    xgb_result = predict_attack_family(
        X,
        model_type="xgboost"
    )

    rf_attack = rf_result["attack_family"]
    xgb_attack = xgb_result["attack_family"]

    agreement = rf_attack == xgb_attack

    if agreement:
        final_attack = rf_attack

        # Conservative confidence:
        # use the lower confidence of the two models
        final_confidence = min(
            rf_result["confidence"],
            xgb_result["confidence"]
        )

        decision = "AGREEMENT"

    else:
        final_attack = None
        final_confidence = None
        decision = "MODEL_DISAGREEMENT"

    return {
        "random_forest": rf_result,
        "xgboost": xgb_result,
        "agreement": agreement,
        "decision": decision,
        "final_attack": final_attack,
        "final_confidence": final_confidence
    }
```

```python
def final_threat_analysis(X):

    comparison = compare_models(X)

    rf_attack = comparison["random_forest"]["attack_family"]
    xgb_attack = comparison["xgboost"]["attack_family"]

    rf_confidence = comparison["random_forest"]["confidence"]
    xgb_confidence = comparison["xgboost"]["confidence"]

    # ------------------------------------------------
    # CASE 1: Models disagree
    # ------------------------------------------------
    if not comparison["agreement"]:

        return {
            "status": "REVIEW_REQUIRED",
            "reason": "MODEL_DISAGREEMENT",

            "random_forest":
                comparison["random_forest"],

            "xgboost":
                comparison["xgboost"],

            "message":
                "The models produced different predictions. "
                "Further investigation is recommended."
        }

    # ------------------------------------------------
    # CASE 2: Models agree on Benign
    # ------------------------------------------------
    if rf_attack == "Benign":

        return {
            "status": "NO_ATTACK_DETECTED",
            "reason": "BOTH_MODELS_PREDICT_BENIGN",

            "final_attack": "Benign",

            "final_confidence": min(
                rf_confidence,
                xgb_confidence
            ),

            "random_forest":
                comparison["random_forest"],

            "xgboost":
                comparison["xgboost"],

            "message":
                "Both models classified the network flow "
                "as benign."
        }

    # ------------------------------------------------
    # CASE 3: Models agree on an attack
    # ------------------------------------------------
    final_attack = rf_attack

    final_confidence = min(
        rf_confidence,
        xgb_confidence
    )

    evidence = get_evidence_for_attack(
        final_attack
    )

    if evidence is None:

        return {
            "status": "REVIEW_REQUIRED",
            "reason": "EVIDENCE_NOT_FOUND",

            "final_attack": final_attack,
            "final_confidence": final_confidence,

            "random_forest":
                comparison["random_forest"],

            "xgboost":
                comparison["xgboost"]
        }

    # Find authoritative source
    source_entry = next(
        (
            source
            for source in authoritative_data["sources"]
            if source["attack_family"].lower()
            == final_attack.lower()
        ),
        None
    )

    source_evidence = {}

    if source_entry:
        source_evidence = source_entry.get(
            "evidence",
            {}
        )

    return {
        "status": "THREAT_CONFIRMED",

        "final_attack": final_attack,

        "final_confidence":
            final_confidence,

        "random_forest":
            comparison["random_forest"],

        "xgboost":
            comparison["xgboost"],

        "source":
            evidence["source_name"],

        "source_url":
            evidence["source_url"],

        "technique_id":
            evidence.get(
                "technique_id",
                ""
            ),

        "technique_name":
            evidence.get(
                "technique_name",
                ""
            ),

        "description":
            source_evidence.get(
                "description",
                ""
            ),

        "indicators":
            source_evidence.get(
                "indicators",
                []
            ),

        "investigation":
            source_evidence.get(
                "detection_guidance",
                []
            ),

        "mitigation":
            source_evidence.get(
                "mitigation_guidance",
                []
            )
    }
```

```python
def final_threat_analysis(X):

    # =========================================================
    # 1. Get predictions from both models
    # =========================================================

    rf_result = predict_attack_family(
        X,
        model_type="random_forest"
    )

    xgb_result = predict_attack_family(
        X,
        model_type="xgboost"
    )

    rf_attack = rf_result["attack_family"]
    xgb_attack = xgb_result["attack_family"]

    rf_confidence = rf_result["confidence"]
    xgb_confidence = xgb_result["confidence"]

    # =========================================================
    # 2. XGBoost is the PRIMARY model
    # =========================================================

    # ---------------------------------------------------------
    # CASE A: XGBoost says BENIGN
    # ---------------------------------------------------------

    if xgb_attack == "Benign":

        # RF also says Benign
        if rf_attack == "Benign":

            return {
                "status": "NO_ATTACK_DETECTED",
                "final_attack": "Benign",

                "final_confidence":
                    xgb_confidence,

                "random_forest":
                    rf_result,

                "xgboost":
                    xgb_result,

                "message":
                    "Both models classified the network flow "
                    "as benign."
            }

        # XGBoost says Benign but RF sees an attack
        else:

            return {
                "status": "REVIEW_REQUIRED",
                "reason": "MODEL_DISAGREEMENT",

                "final_attack": "Benign",

                "final_confidence":
                    xgb_confidence,

                "random_forest":
                    rf_result,

                "xgboost":
                    xgb_result,

                "message":
                    "XGBoost classified the flow as benign, "
                    "but Random Forest detected an attack. "
                    "Further investigation is recommended."
            }

    # =========================================================
    # 3. XGBoost detected an attack
    # =========================================================

    final_attack = xgb_attack

    # ---------------------------------------------------------
    # CASE B: Both models agree on the attack
    # ---------------------------------------------------------

    if rf_attack == xgb_attack:

        status = "THREAT_CONFIRMED"

        final_confidence = min(
            rf_confidence,
            xgb_confidence
        )

    # ---------------------------------------------------------
    # CASE C: XGBoost detects attack but RF disagrees
    # ---------------------------------------------------------

    else:

        return {
            "status": "REVIEW_REQUIRED",
            "reason": "MODEL_DISAGREEMENT",

            "final_attack": xgb_attack,

            "final_confidence":
                xgb_confidence,

            "random_forest":
                rf_result,

            "xgboost":
                xgb_result,

            "message":
                "XGBoost detected an attack, but "
                "Random Forest produced a different "
                "prediction. Further investigation "
                "is recommended."
        }

    # =========================================================
    # 4. Retrieve authoritative evidence
    # =========================================================

    evidence = get_evidence_for_attack(
        final_attack
    )

    if evidence is None:

        return {
            "status": "REVIEW_REQUIRED",
            "reason": "EVIDENCE_NOT_FOUND",

            "final_attack":
                final_attack,

            "final_confidence":
                final_confidence,

            "random_forest":
                rf_result,

            "xgboost":
                xgb_result
        }

    # =========================================================
    # 5. Find authoritative knowledge entry
    # =========================================================

    source_entry = next(
        (
            source
            for source in authoritative_data["sources"]
            if source["attack_family"].lower()
            == final_attack.lower()
        ),
        None
    )

    source_evidence = {}

    if source_entry:

        source_evidence = source_entry.get(
            "evidence",
            {}
        )

    # =========================================================
    # 6. Return complete threat analysis
    # =========================================================

    return {

        "status":
            "THREAT_CONFIRMED",

        "final_attack":
            final_attack,

        "final_confidence":
            final_confidence,

        "random_forest":
            rf_result,

        "xgboost":
            xgb_result,

        "source":
            evidence["source_name"],

        "source_url":
            evidence["source_url"],

        "technique_id":
            evidence.get(
                "technique_id",
                ""
            ),

        "technique_name":
            evidence.get(
                "technique_name",
                ""
            ),

        "description":
            source_evidence.get(
                "description",
                ""
            ),

        "indicators":
            source_evidence.get(
                "indicators",
                []
            ),

        "investigation":
            source_evidence.get(
                "detection_guidance",
                []
            ),

        "mitigation":
            source_evidence.get(
                "mitigation_guidance",
                []
            )
    }
```

```python
primary_results = []

SAMPLES_PER_CLASS = 100

for expected_family, (filename, valid_labels) in evaluation_samples.items():

    print(f"Processing {expected_family}...")

    path = os.path.join(
        dataset_path,
        filename
    )

    df = pd.read_parquet(path)

    class_rows = df[
        df["Label"].isin(valid_labels)
    ]

    sampled_rows = class_rows.sample(
        n=min(SAMPLES_PER_CLASS, len(class_rows)),
        random_state=42
    )

    for _, row in sampled_rows.iterrows():

        actual_dataset_label = row["Label"]

        # Get the original 77 features
        X = row.drop(
            labels=["Label"]
        ).to_frame().T

        # Make sure all features are numeric
        X = X.apply(
            pd.to_numeric,
            errors="coerce"
        )

        # Skip invalid rows
        if X.isna().any().any():
            continue

        # Run the NEW XGBoost-primary decision engine
        result = final_threat_analysis(X)

        primary_results.append({

            "expected_family":
                expected_family,

            "dataset_label":
                actual_dataset_label,

            "status":
                result["status"],

            "final_attack":
                result.get(
                    "final_attack",
                    ""
                ),

            "final_confidence":
                result.get(
                    "final_confidence",
                    None
                ),

            "rf_prediction":
                result["random_forest"][
                    "attack_family"
                ],

            "xgb_prediction":
                result["xgboost"][
                    "attack_family"
                ]
        })


primary_results_df = pd.DataFrame(
    primary_results
)

print("\n" + "=" * 70)
print("NEW DECISION ENGINE EVALUATION COMPLETED")
print("=" * 70)

print(
    "Total samples:",
    len(primary_results_df)
)

print("\nStatus counts:")

print(
    primary_results_df[
        "status"
    ].value_counts()
)
```
